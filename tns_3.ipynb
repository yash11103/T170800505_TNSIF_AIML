```python
#7 Reverse a number using a loop

num = int(input("enter a number"))
reverse = 0

while num!= 0:
  digit = num % 10
  reverse = reverse * 10 + digit
  num = num // 10
print(reverse)
```

```python
#8 Check whether a number is prime.
num = int(input("enter a number"))
count = 0

for i in range(1, num+1):
  if num % i == 0:
    count = count + 1
if count == 2:
  print(num, "is prime")
else:
  print(num, "is not prime")
```

```python
#9 Write a function to check if a number is palindrome

n = input(" enter a value:")
if n == n[::-1]:
  print("is palnidroem  ")
else:
  print("not")
```

```python
#10  Write a function to calculate sum of digits of a number

def sum_of_digits(num):
    total = 0
    while num != 0:
        digit = num % 10
        total = total + digit
        num = num // 10
    return total

n = int(input("Enter a number: "))
print("Sum of digits:", sum_of_digits(n))
```

```python
#25 Sort dictionary by values.

students = {
    "Amit": 85,
    "Riya": 92,
    "John": 78,
    "Sara": 90
}

items = __builtins__.list(students.items())

for i in range(len(items)):
    for j in range(len(items) - 1):
        if items[j][1] > items[j + 1][1]:
            items[j], items[j + 1] = items[j + 1], items[j]

sorted_dict = dict(items)

print("Sorted by values:", sorted_dict)
```

```python
#27 Create a class Rectangle to calculate area and perimeter.

class Rectangle:
  def __init__ (self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * (self.length + self.width)

  r1 = Rectangle(10, 5)
  print(r1.area())
  print(r1.perimeter())

```

```python
#28 Create a class Employee with salary and method to increase salary.
class Employee:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def increase_salary(self, amount):
    self.salary = self.salary + amount

e1 = Employee("sidd", 30000)
print(e1.name)
print(e1.salary)

e1.increase_salary(4000)
print(e1.salary)
```

```python
#29 Create a class ShoppingCart using dictionary (item: price) and calculate total bill

class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def total_bill(self):
        return sum(self.items.values())

cart = ShoppingCart()
cart.add_item("Bread", 40)
cart.add_item("Milk", 25)
cart.add_item("Eggs", 60)

print("Cart Items:", cart.items)
print("Total Bill:", cart.total_bill())
```

```python
#30 Create a class Library to store books in a list and add/remove books.

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        self.books.remove(book)

lib = Library()

lib.add_book("Harry Potter")
lib.add_book("Python Basics")
lib.add_book("The Alchemist")

print("Books in Library:", lib.books)

lib.remove_book("Python Basics")

print("Books after removing one:", lib.books)
```
