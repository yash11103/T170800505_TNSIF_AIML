What is Data Cleaning?
Data Cleaning is the process of:

Fixing incorrect data

Handling missing values

Removing duplicates

Standardizing formats

🔹 Why is Data Preprocessing Important?
Machine Learning models:
Cannot handle missing or messy data properly

Require numerical and consistent input

🔹 Example of Messy Data:

Name	Age	City

Ram	25	Mumbai

Shyam	NaN	mumbai

Ram	25	Mumbai



Problems:
Missing values

Inconsistent text

Duplicate rows

```python
data = {
    "Name": ["Ram", "Shyam", "Ram", "Sita", None,"Shyam","Shyam","Shyam","Ram"],
    "Age": [25, np.nan, 25, 30, 22,13,17,18,19],
    "City": ["Mumbai", "mumbai", "Mumbai", "Delhi", "Delhi", "Delhi", "Delhi", "Delhi", "Delhi"]
}
print(data)
df = pd.DataFrame(data)
df
```

Common Data Issues
Missing Values → NaN

Duplicate Records

Incorrect Data Types

Inconsistent Text (Mumbai vs mumbai)

We must identify and fix these issues before using data.

## 🔹 Missing Values

Missing values are represented as:
- NaN (Not a Number)
- None

---

## 🔍 Steps:
1. Identify missing values
2. Count them
3. Decide:
   - Drop
   - Fill

```python
# Check missing values
df.isnull()

# Count missing values
df.isnull().sum()
```

```python
# Drop rows with missing values
df_drop = df.dropna()
df_drop
```

## 🔹 Filling Missing Values

### Numerical Columns:
- Mean
- Median

### Categorical Columns:
- Mode

---

## 🧠 When to Use:
- Mean → Normal data
- Median → Skewed data
- Mode → Categories

```python
# ✅ Best Practice

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Name"] = df["Name"].fillna(df["Name"].mode()[0])


df
```

## 🔹 Selecting Multiple Columns

To work on multiple columns together:

### ✅ Syntax:
df[["col1", "col2"]]

---

## 🔹 Why use this?
- Apply same operation on multiple columns
- Clean multiple columns at once
- Save time and avoid repetition

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, np.nan, 25, 30,20],
    "Salary": [20000, 25000, np.nan, 40000,20000],
    "City": ["Nashik", "Delhi", "Pune", "Akola","Mumbai"]
})

df
```

```python
# Fill multiple columns with mean

df[["Age", "Salary"]] = df[["Age", "Salary"]].fillna(df[["Age", "Salary"]].mean())

df
```

```python
# Convert multiple columns to integer

df[["Age", "Salary"]] = df[["Age", "Salary"]].astype(int)

df.dtypes
df
```

```python
# Increase values by 10%

df[["Age", "Salary"]] = df[["Age", "Salary"]].apply(lambda x: x * 1.1)

df
```

```python
df
```

```python
df
```

```python
# Replace specific values

df[["City"]] = df[["City"]].replace("Mumbai", "Pune")

df
```

## 🔹 Handle Missing Values

### Numerical Columns:
- rainfall
- temperature
- humidity
- wind_speed

➡️ Fill with mean

### Categorical:
- weather_condition

➡️ Fill with mode

```python
import pandas as pd

# Numerical columns
num_cols = ["rainfall", "temperature", "humidity", "wind_speed"]

# Fill numerical missing values with mean
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

# Fill categorical missing values with mode
df["weather_condition"] = df["weather_condition"].fillna(df["weather_condition"].mode()[0])

# Check missing values
print(df.isnull().sum())
```

# 🎓 Student Performance Dataset – Data Cleaning

## 🎯 Objective:
Clean and preprocess student dataset for ML

---

## 🔹 Columns:
- name
- age
- gender
- marks
- attendance
- city

---

## ⚠️ Possible Issues:
- Missing values
- Inconsistent text (Male / male)
- Invalid marks (>100 or <0)
- Duplicates

```python

import pandas as pd
import numpy as np

data = {
    "name": ["Amit", "Sara", "John", "Amit", None],
    "age": [20, 21, np.nan, 20, 19],
    "gender": ["Male", "female", "Male", "Male", "FEMALE"],
    "marks": [85, 90, 105, 85, -5],
    "attendance": [90, 85, 80, 90, None],
    "city": ["Mumbai", "Delhi", "mumbai", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
df.isnull().sum()
df.describe()
```

## 🔹 Handling Missing Values

- age → mean
- attendance → mean
- name → "Unknown"

```python
df["age"].fillna(df["age"].mean(), inplace=True)
df["attendance"].fillna(df["attendance"].mean(), inplace=True)
df["name"].fillna("Unknown", inplace=True)

df
```

```python
df["gender"] = df["gender"].str.lower()
df["city"] = df["city"].str.lower()

df
```

## 🔹 Fix Invalid Data

Marks should be:
- Between 0 and 100

```python
df = df[(df["marks"] >= 0) & (df["marks"] <= 100)]

df
```

```python
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].mean())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Removed {initial_rows - len(df)} duplicate rows.")
```

```python
df['Date_of_Birth'] = pd.to_datetime(df['Date_of_Birth'], errors='coerce')
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'], errors='coerce')
```

```python
df.columns = [col.lower() for col in df.columns]
print("Updated Column Names:", df.columns.tolist())
```

```python
for col in df.columns:
    if df[col].dtype == 'float64':
        df[col] = df[col].astype(int)

print("\nFinal Data Types:\n", df.dtypes)
```

```python
import pandas as pd

df = pd.read_csv('data.csv')

# Select multiple specific columns and fill their missing values with 0 (or their respective mean)
cols_to_fill = ['Math_Marks', 'Science_Marks', 'English_Marks']
df[cols_to_fill] = df[cols_to_fill].fillna(df[cols_to_fill].mean())
```

```python
# Increase math marks by 10%
df['Math_Marks'] = df['Math_Marks'] * 1.10
```

```python
# Convert to title case to fix capitalization mismatches
df['Department'] = df['Department'].str.title()

# Fix variations like 'It' to 'Information Technology' if needed
df['Department'] = df['Department'].replace({'It': 'Information Technology'})
```

```python
# Calculate average marks across subjects
df['Average_Marks'] = df[['Math_Marks', 'Science_Marks', 'English_Marks']].mean(axis=1)

# Assign Pass if average >= 50, else Fail
df['Pass_Fail'] = df['Average_Marks'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')
```

```python
# Filter students with Math marks greater than 80
high_math_scorers = df[df['Math_Marks'] > 80]

# Example for rainfall (if rainfall column exists in another dataset):
# high_rainfall = df[df['Rainfall'] > 10]
```

```python
# Drop the 'Email' column
df = df.drop(columns=['Email'])
```

```python
# Apply one-hot encoding on categorical columns like Gender and Department
df_encoded = pd.get_dummies(df, columns=['Gender', 'Department'])
```
